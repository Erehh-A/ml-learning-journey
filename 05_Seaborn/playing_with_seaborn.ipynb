```python
import seaborn as sns
import pandas as pd
# sns.get_dataset_names()
```

```python
pengs = sns.load_dataset('penguins')
print(pengs)
```

### Scatter Plot

```python
sns.scatterplot(data=pengs, x='body_mass_g', y='flipper_length_mm', hue='sex')
```

### Styling

```python
# sns.set_style('whitegrid') 
sns.set_style('darkgrid') 
sns.scatterplot(data=pengs, x='body_mass_g', y='flipper_length_mm', hue='sex', palette='Set1')
```

### Histogram

```python
sns.histplot(data=pengs, x='body_mass_g', hue='sex', multiple='stack') # Multiple : Better understaning in this graph
```

### Joint plot


```python
sns.jointplot(data=pengs, x='body_mass_g', y='flipper_length_mm')
# The histograms shows the density
```

### Count Plot

```python
sns.countplot(data=pengs, x='island', hue='sex', palette='Set2')
# Used to count
```

### Box Plot
upper quartile = 75%ile
median = 50%ile
lower quartile = 25%ile

<img src="img.png" style="width:350px;height:350px">

```python
sns.boxenplot(data=pengs, x='island', y='body_mass_g', hue='sex')
```

### Violine plot
<img src="img2.png" style="width:350px;height:350px">

```python
sns.violinplot(data=pengs, x='island', y='body_mass_g', hue='sex')
# sns.violinplot(data=pengs, x='island', y='body_mass_g', hue='sex', split=True) 
# split attribute combines the graph as a single 
```

### KDE Plot

```python
# Line version of histogram (Kernel Density Estimate)
sns.kdeplot(data=pengs, x='body_mass_g', hue='sex', fill=True)
# fill attribute is used to fill the color inside the graph
```

### Heatmap
<img src ='img3.png'>

```python
columns = ["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"]
pengs[columns].corr()
```

```python
sns.heatmap(data=pengs[columns].corr(), annot=True, cmap='Reds', linewidths=2, linecolor='Black')
```

### Pair Plot

```python
# Relation between all of the numerical data
sns.pairplot(data=pengs, hue='sex', palette='Set2')
```
