```python
import pandas as pd 
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv('cars.csv')
df.sample(10)
```

```python
df[['owner']].value_counts()
```

### OneHotEncoding using Sklearn

```python
x_train, x_test, y_train, y_test = train_test_split(df.iloc[:, :4], df.iloc[:, -1], test_size=0.2, random_state=10)
```

```python
ohe = OneHotEncoder(drop='first', sparse_output=False, dtype=np.int32)
```

```python
x_train_new = ohe.fit_transform(x_train[['fuel', 'owner']])
x_test_new = ohe.transform(x_test[['fuel', 'owner']])

x_train_new.shape
```

```python
np.hstack((x_train[['brand', 'km_driven']].values, x_train_new))
```

### OneHotEncoding with Top Categories

```python
counts = df['brand'].value_counts()
# df['brand'].nunique()
threshold = 100
replace = counts[counts <= threshold].index
pd.get_dummies(df['brand'].replace(replace, 'uncommon'), dtype=np.int32).sample(5)
```
