### Transpose (to change the dimensions of a matrix)

```python
t1 = torch.tensor([[1, 2, 4], 
                  [3, 5, 7]])

t2 = torch.tensor([[2, 6, 7],
                  [6, 2, 6]])

print(f"Origina tensors shape are : {t1.shape} and {t2.shape}, which can't get multiply")
print("Shape after taking the transpose of the second tensor : \n")

print(f"{t1.shape} and {t2.T.shape}")

print("\nFinal Output")
print(f"{torch.matmul(t1, t2.T)}")
```

### Reshaping, stacking, squeezing and unsqueezing

```python
import torch 
x = torch.arange(1., 7.)
x, x.shape 
```

```python
# Reshaping (Reshapes input to shape (if compatible), can also use torch.Tensor.reshape().)
x_reshaped = x.reshape(1, 6)
x_reshaped, x_reshaped.shape
```

```python
# View (Returns a view of the original tensor in a different shape but shares the same data as the original tensor.)
z = x.view(3, 2)
# z, z.shape 
```

```python
# Stacking (Concatenates a sequence of tensors along a new dimension (dim), all tensors must be same size)
s = torch.stack([x, x, x])
s
```

```python
# Squeezing (Squeezes input to remove all the dimenions with value 1)
print(f"Before squeezing x_reshaped : {x_reshaped} and shape : {x_reshaped.shape}") 
x_reshaped_squeezed = x_reshaped.squeeze()
print(f"After squeezing x_reshaped : {x_reshaped_squeezed} and shape : {x_reshaped_squeezed.shape}")
```

```python
# Unsqueezing (Opposite of squeezing)
print(f"Afer squeezing : x_reshaped : {x_reshaped_squeezed} and shape : {x_reshaped_squeezed.shape}")
x_unsqueezed = x_reshaped_squeezed.unsqueeze(dim=0)
print(f"After unsqueezing, x_rechaped : {x_unsqueezed} and shape : {x_unsqueezed.shape}")
```

```python
# Permuting (arranging) (Returns a view of the original input with its dimensions permuted (rearranged) to dims)
import torch 
x_original = torch.rand(size=(223, 34, 2))
print(f"Original shape is : {x_original.shape}")

x_permuted = x_original.permute(2, 0, 1)   # 2 to 0, 0 to 1 and 1 to 2
print(f"After permuted : {x_permuted.shape}")
```

### Running tensors on GPUs (and making faster computations)

```python
# !nvidia-smi
```

```python
# Check for GPU
import torch 
torch.cuda.is_available()
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu" 
device 
```

```python
tensor = torch.arange(1, 6)
tensor, tensor.device 
```

```python
tensor_to_gpu = tensor.to(device) 
tensor_to_gpu.device 
```

```python
temp = tensor_to_gpu * 9998989
print(f"Original tensor running on gpu : {tensor_to_gpu}")
print(f"Multiplying by a large number : {temp}")
```

```python
# Workin with numpy (numpy can't run in gpu)
tensor_to_cpu = tensor_to_gpu.cpu().numpy() 
tensor_to_cpu, tensor_to_cpu * 1434
```
