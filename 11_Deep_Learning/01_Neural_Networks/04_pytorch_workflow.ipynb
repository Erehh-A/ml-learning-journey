```python
import torch 
from torch import nn 
from matplotlib import pyplot as plt 
```

```python
# Creating known parameters 
weight = 0.7
bias = 0.2 

start = 0
end = 1
step = 0.02

X = torch.arange(start, end, step).unsqueeze(dim=1)
Y = weight * X + bias   # Linear regression for example 
X[: 10], X.shape   # Initial predictions 
```

```python
Y[: 10], Y.shape    # Desired output
```

### Splitting data into training and test sets 

```python
train_split = int((0.8 * len(X)))
X_train, Y_train = X[:train_split], Y[:train_split]
X_test, Y_test = X[train_split:], Y[train_split:]

len(X_train), len(Y_train), len(X_test), len(Y_test)
```

```python
# Visualising the data
def plot_prediction(train_data = X_train,
                    train_label = Y_train,
                    test_data = X_test,
                    test_label = Y_test,
                    predictions = None) :
    plt.figure(figsize=(7, 7))     # Length, bredth

    # Plotting training data
    plt.scatter(train_data, train_label, c="b", s=7, label="Training Data") 

    # Plotting test data
    plt.scatter(test_data, test_label, c="r", s=7, label="Testing Data")

    # Plotting predictions IF exists
    if predictions is not None:
        # Predictions will be made on the test data
        plt.scatter(test_data, predictions, c="g", s=7, label="Predictions")

    plt.legend()
    plt.show()
```

```python
plot_prediction()
```
